##  Job Role-wise Employee Distribution

```python
plt.figure(figsize=(10, 5))
sns.countplot(data=df, y="JobRole", order=df["JobRole"].value_counts().index)
plt.title("Employee Distribution by Job Role")
plt.xlabel("Number of Employees")
plt.ylabel("Job Role")
plt.show()
```

##### The Laboratory Technician and Research Scientist roles have relatively high employee counts. Sales Representative and Human Resources roles have comparatively fewer employees.

## Job Role-wise Employee Attrition

```python
plt.figure(figsize=(11, 6))
sns.countplot(
    data=df,
    y="JobRole",
    hue="Attrition"
)

plt.title("Employee Attrition by Job Role")
plt.xlabel("Number of Employees")
plt.ylabel("Job Role")
plt.legend(title="Attrition")
plt.show()
```

##### Sales Representatives have a particularly high number of employees leaving relative to the size of their role. Research Scientists and Laboratory Technicians also contribute to employee attrition.

## Age Group-wise Employee Distribution

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Age", bins=15, kde=True)
plt.title("Employee Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.show()
```

##### The chart shows the age distribution of employees. Most employees fall within the younger and middle-age groups, with fewer employees in the older age ranges.

## Age Group-wise Employee Attrition

```python
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 25, 35, 45, 55, 100],
    labels=["Under 25", "25-34", "35-44", "45-54", "55+"]
)
plt.figure(figsize=(9, 5))
sns.countplot(data=df, x="AgeGroup", hue="Attrition")
plt.title("Employee Attrition by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Employees")
plt.legend(title="Attrition")
plt.show()
```

##### The 25–34 age group has the highest number of employees leaving the company. Attrition counts are comparatively lower in the older age groups.

## Overtime-wise Employee Attrition

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="OverTime", hue="Attrition")
plt.title("Employee Attrition by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Number of Employees")
plt.legend(title="Attrition")
plt.show()
```

##### This chart compares employees who work overtime with those who do not. It shows the number of employees who stayed and left in each group.

## Attrition Rate by Overtime

```python
overtime_rate = (
    df.groupby("OverTime")["Attrition"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .reset_index(name="Attrition Rate")
)
plt.figure(figsize=(7, 5))
sns.barplot(data=overtime_rate, x="OverTime", y="Attrition Rate")
plt.title("Attrition Rate by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Attrition Rate (%)")
plt.show()

overtime_rate
```

##### This chart compares the percentage of employees leaving in the two overtime groups. A higher percentage indicates greater attrition within that group.

## Multivariate Analysis: Age Group and Overtime

```python
df["AttritionFlag"] = (df["Attrition"] == "Yes").astype(int)
multi_analysis = (
    df.groupby(["AgeGroup", "OverTime"], observed=True)["AttritionFlag"]
    .mean() * 100
)
multi_analysis = multi_analysis.unstack()
plt.figure(figsize=(8, 5))
sns.heatmap(
    multi_analysis,
    annot=True,
    fmt=".2f",
    cmap="YlOrRd"
)
plt.title("Attrition Rate by Age Group and Overtime")
plt.xlabel("Overtime")
plt.ylabel("Age Group")
plt.show()
```

##### The heatmap compares attrition percentages across different age groups and overtime categories. Darker cells represent higher attrition rates.

## Outlier Analysis: Monthly Income

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="MonthlyIncome")
plt.title("Outlier Analysis of Monthly Income")
plt.xlabel("Monthly Income")
plt.show()
```

##### The boxplot displays the distribution of employee monthly income and highlights unusually high or low values. Points beyond the whiskers are potential outliers.

## Outlier Detection Using IQR Method

```python
Q1 = df["MonthlyIncome"].quantile(0.25)
Q3 = df["MonthlyIncome"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["MonthlyIncome"] < lower_limit) |
    (df["MonthlyIncome"] > upper_limit)
]

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Number of Outliers:", len(outliers))
```

##### The IQR method identifies unusually low or high monthly income values. These are potential outliers and should be investigated rather than automatically removed.

## Correlation Analysis

```python
numerical_data = df.select_dtypes(include="number")
correlation_matrix = numerical_data.corr()
plt.figure(figsize=(14, 10))
sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    annot=False,
    linewidths=0.5
)
plt.title("Correlation Heatmap of Numerical Variables")
plt.show()
```

##### The heatmap shows the strength and direction of relationships between numerical variables. Positive values indicate variables tend to increase together, while negative values indicate an inverse relationship.

## Department and Attrition Association

```python
department_attrition = pd.crosstab(
    df["Department"],
    df["Attrition"],
    normalize="index"
) * 100

plt.figure(figsize=(8, 5))

sns.heatmap(
    department_attrition,
    annot=True,
    fmt=".2f",
    cmap="Blues"
)

plt.title("Department-wise Attrition Percentage")
plt.xlabel("Attrition Status")
plt.ylabel("Department")

plt.show()

department_attrition
```

##### The heatmap compares the percentage of employees who stayed and left in each department. It helps identify departments with relatively higher attrition.

## Attrition Rate by Job Role

```python
jobrole_attrition = (
    df.groupby("JobRole")["AttritionFlag"]
    .mean()
    .mul(100)
    .reset_index(name="Attrition Rate")
    .sort_values("Attrition Rate", ascending=False)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=jobrole_attrition,
    x="Attrition Rate",
    y="JobRole"
)

plt.title("Attrition Rate by Job Role")
plt.xlabel("Attrition Rate (%)")
plt.ylabel("Job Role")

plt.show()

jobrole_attrition
```

##### The chart ranks job roles according to their employee attrition rates. It helps identify which roles have relatively higher employee turnover.

## Key Business Insights

```python
print("Total Employees:", len(df))

print("Employees Left:", (df["Attrition"] == "Yes").sum())

print("Overall Attrition Rate:",
      round((df["Attrition"] == "Yes").mean() * 100, 2), "%")

print("\nDepartment-wise Attrition Rate:")
print(
    (pd.crosstab(df["Department"], df["Attrition"], normalize="index")["Yes"] * 100)
    .round(2)
)

print("\nJob Role-wise Attrition Rate:")
print(jobrole_attrition.round(2).to_string(index=False))

print("\nOvertime-wise Attrition Rate:")
print(overtime_rate.round(2).to_string(index=False))
```

##### These findings help HR identify departments and job roles with higher employee attrition. They can support decisions about employee satisfaction, workload management and retention strategies.
